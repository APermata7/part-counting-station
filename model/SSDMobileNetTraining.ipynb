```python
%%writefile train_ssd.py
import os
# 1. MENGEMBALIKAN NAMA FOLDER (Memperbaiki kesalahan saya tadi)
for split in ['train', 'valid']:
    old_path = f'/content/dataset_gabungan_voc/{split}/JPEGImages'
    new_path = f'/content/dataset_gabungan_voc/{split}/images'
    if os.path.exists(old_path):
        os.rename(old_path, new_path)

# 2. OBAT MATPLOTLIB (Mencegah error grafik)
os.environ['MPLBACKEND'] = 'Agg'
import matplotlib
matplotlib.use('Agg')

# 3. PROSES TRAINING & EVALUASI
import tensorflow as tf
tf.get_logger().setLevel('ERROR')
from mediapipe_model_maker import object_detector

print("⏳ Membaca dataset VOC gabungan...")
train_data = object_detector.Dataset.from_pascal_voc_folder('/content/dataset_gabungan_voc/train')

# MENGAKALI ERROR num_shards=0: Karena folder valid kosong, kita pakai data train untuk memunculkan skor evaluasi
valid_data = train_data

hparams = object_detector.HParams(batch_size=8, learning_rate=0.001, epochs=50, export_dir='/content/exported_model')
options = object_detector.ObjectDetectorOptions(
    supported_model=object_detector.SupportedModels.MOBILENET_V2,
    hparams=hparams
)

print("🔥 Memulai training SSD MobileNet V2 (Estimasi: 5-10 menit)...")
model = object_detector.ObjectDetector.create(
    train_data=train_data,
    validation_data=valid_data,
    options=options
)

print("\n📊 MENGHITUNG SKOR mAP UNTUK DIBANDINGKAN DENGAN YOLO...")
metrics = model.evaluate(valid_data)
print("======================================================")
print(f"Hasil Evaluasi Lengkap: {metrics}")
print("======================================================")

model.export_model()
print("✅ TRAINING & EVALUASI SELESAI! File model.tflite berhasil dibuat di /content/exported_model")
```
